```python
# Your code here
#1. Now I read another excel file "Steps.xlsx" into a pandas dataframe
all_sheets = pd.read_excel("Steps.xlsx", sheet_name=None)

print(f"Keys in all_sheets: {list(all_sheets.keys())}")
```

```python
# Your code here

#2. Now I will concatenate all the sheets into a single dataframe
df = pd.concat(all_sheets.values(), ignore_index=True)

df
```

```python
# Your code here

#3. Now I will extract the year, month, and day from the "Date" column and create new columns for each
# Note: I refer this way, cleanest and professional way
# a) Practice for pd.concat() function
# b) Practice for pd.DataFrame() function
df = pd.concat ([df, pd.DataFrame({
    "Year": df["Date"].dt.year,
    "Month": df["Date"].dt.month_name().str[:3],
    "Day": df["Date"].dt.day
})], axis=1)
df
```

```python
# Your code here

#4. Make boxplot where there is for each month amount of steps separated by year
# Boxplot: steps by month, with a separate box per year
plt.figure(figsize=(12,6))
sns.boxplot(data=df, x="Month", y="Steps", hue="Year", palette="Set2")
plt.title("Steps by Month (separated by Year)")
plt.xlabel("Month")
plt.ylabel("Steps")
plt.legend(title="Year")
plt.show()
```
