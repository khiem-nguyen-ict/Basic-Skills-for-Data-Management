### Task 8: Explain the code and plot. 
- Use the given code, no need to change it
- Figure out, what does correlation mean.
- Explain the code.
- Figure out, how to read the heatmap. Why is there always 1 (or 1.00) on the diagonal? What do the colors indicate?

```python
corr = penguins.select_dtypes("number").corr()

# corr is the correlation between the NUMBER columns from penguin data frame only.

fig, ax = plt.subplots(figsize=(8, 6))

# The heatmap shows the correlation between each pair of columns.
# A value of 1 means a 100% positive correlation, while 0 indicates no relationship.
# For example, a column correlated with itself always has a value of 1 (100%), closed to Red color.
# Otherwise, its color is closed to blue (see the legend color bar on the right side of the image)
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

ax.set_title("Correlation Matrix")

plt.show()
```
