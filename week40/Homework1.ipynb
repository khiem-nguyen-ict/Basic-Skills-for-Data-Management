## Task 2: Explain the code (add explanation as comments to the code cells) (1 point)
- Do not change the code, only add comments.

```python
# Explain this code

cols = df.columns
print(cols)

# This code gets and prints the column names (headers) of the DataFrame
```

```python
# Explain this code

for col in cols:
    df[col] = df[col].apply(lambda x: x.replace(',', '.') if isinstance(x, str) else x)

# This code iterates through each column in the DataFrame and replaces the 
# Finnish decimal separator (',') with the US separator ('.') for string values.
# Non-string values remain unchanged.

# My takeaway: It scans all cells in the DataFrame to apply the formatting magic, 
# but there is an invisible bug: if other columns (such as an address) contain commas, 
# this approach will break those values.

# Consider to select a collection of specific columns only to avoid this bug!

df.head(10)

# This code print top 10 rows of the DataFrame to check
```

### Let's do some conversion here.

```python
# Convert data types to numeric in selected columns.
# Do not change this code.
cols = ['Distance', 'Calories', 'Avg HR', 'Max HR', 'Aerobic TE', 'Avg Cadence', 'Max Cadence',
       'Avg Pace', 'Best Pace', 'Total Ascent', 'Total Descent',
       'Avg Stride Length', 'Avg Cadence.1', 'Max Cadence.1',
       'Training Stress Score®', 'Total Strokes', 'Avg. Swolf',
       'Avg Stroke Rate', 'Steps', 'Min Temp', 'Decompression',
       'Number of Laps', 'Max Temp', 'Min Elevation', 'Max Elevation']
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')
df.info()

# My reflection: Targeting specific columns is the best way to avoid the previous bug.
# My takeaway: Pandas also has its own utility function, pd.to_numeric, to convert values to numbers. Cool!
```

## Task 3: Why is there suddenly now a lot of missing values? (1 point)

The reason there are a lot of missing values is that `errors='coerce'` in `df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')` replaces invalid numeric strings with `NaN`.

As a result, only valid numbers remain.

## Task 4: Remove all the activities where the distance is less than 5 (km) (1 point)

```python
# Keep rows where Distance is greater than or equal to 5
df = df[df["Distance"] >= 5]
# or df = df.drop(df[df["Distance"] < 5].index)
df
```

## Task 5: Convert "Date" column into datetime format and check with info() what happened (1 point)

```python
df["Date"] = df["Date"].apply(pd.to_datetime)
df.info()

# Before this, the column Date is object
# After this, the cilumn Date is datetime64[ns]

# ==> Data Type of the column Date has been changed
```

## Task 6: Remove columns that do not seem to be useful (1 point)
- Drop columns containing a lot of missing values
- Drop columns that seem irrelevant

```python
# Run this code to list out columns which have many NaN (missing value)
missing_counts = df.isna().sum()

cols_over_10_missing = missing_counts[missing_counts > 10].index

print(f"List of columns that have many NaN values", cols_over_10_missing)

# drop columns
df.drop(columns=cols_over_10_missing, inplace=True)

df
```

```python
# Next step, show df's describe again

df.describe()

# With the data showing up, the column "Training Stress Score®" does not change. Its value is only 0.
# Drop all columns that min == max

min_vals = df.min(numeric_only=True)
max_vals = df.max(numeric_only=True)

constant_cols = min_vals[min_vals == max_vals].index.tolist()
df.drop(columns=constant_cols, inplace=True)

df
```

## Tasks 7-10: Visualize data (4 points, 1 per plot)
- Create at least 4 different figures using the data (one histogram, one bar plot, one box plot and one scatter plot (where data is separated by activity type)). Use approrpriate titles, axis names, labels, colors, etc. (= plots are readable and understandable).
- Use Matplotlib and Seaborn.
- Find out first, which type of data is best presented with above mentioned plots!



```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=df, x="Distance", ax=axes[0], color="skyblue", kde=True)
axes[0].set_title("Distance Distribution")

sns.histplot(data=df, x="Calories", ax=axes[1], color="salmon", kde=True)
axes[1].set_title("Calories Distribution")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
g = df.groupby("Activity Type")
activity_totals = g["Distance"].sum()
categories_totals = g["Calories"].sum()

axes[0].bar(x=activity_totals.index, height=activity_totals.values, color="green")
axes[0].set_xlabel("Activity Type")
axes[0].set_ylabel("Total Distance")
axes[0].set_title("Total Distance by Activity Type")
axes[0].tick_params(axis="x", labelsize=8, rotation=45)

axes[1].bar(x=categories_totals.index, height=categories_totals.values, color="red")
axes[1].set_xlabel("Activity Type")
axes[1].set_ylabel("Calories")
axes[1].set_title("Total Calories by Activity Type")
axes[1].tick_params(axis="x", labelsize=8, rotation=45)


plt.tight_layout()
plt.show()
```

```python
# Grouped by a categorical column
axes = df.boxplot(column=["Distance", "Calories"], by="Activity Type", figsize=(12, 6))
plt.title("Distance by Activity Type")
plt.suptitle("")
plt.xlabel("Activity Type")
plt.ylabel("Distance")
plt.xticks(rotation=45)
for ax in axes.flatten():
    ax.tick_params(axis="x", labelsize=8, labelrotation=45)
    
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
g = df.groupby("Activity Type")
activity_totals = g["Distance"].sum()
categories_totals = g["Calories"].sum()

axes[0].scatter(x=activity_totals.index, y=activity_totals.values, color="blue")
axes[0].set_xlabel("Activity Type")
axes[0].set_ylabel("Total Distance")
axes[0].set_title("Total Distance by Activity Type")
axes[0].tick_params(axis="x", labelsize=8, rotation=45)

axes[1].scatter(x=categories_totals.index, y=categories_totals.values, color="violet")
axes[1].set_xlabel("Activity Type")
axes[1].set_ylabel("Total Calories")
axes[1].set_title("Total Calories by Activity Type")
axes[1].tick_params(axis="x", labelsize=8, rotation=45)

plt.tight_layout()
plt.show()
```
