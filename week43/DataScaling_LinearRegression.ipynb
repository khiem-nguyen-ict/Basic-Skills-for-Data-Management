### Data Standardization
- Data is converted to so called z-scores, where each value z = (x - mean)/standard deviation
- This can be done using StandardScaler from the Scikit-Learn library
- fit_transform completes the conversion for the StandardScaler object

```python
# data standardization with  sklearn
from sklearn.preprocessing import StandardScaler
stdScaler = StandardScaler()
scaler = stdScaler.fit_transform(df)
# Note data is transposed to get it presented in less space in this notebook
print(scaler.transpose())
```

```python
# Values are
print("shape of scaler:", scaler.shape)
print("data as it is:")
scaler
```

```python
# Create Serie where standardised data is included
stand_rent = scaler[:,0]
stand_rent
```

#### Plot standardized data using Histogram
- Note that the data shape with histogram is different to the original data.
- Bins = 30 is used just for the example

```python
# plot scaled data using histogram
sns.histplot(data = stand_rent ,bins=30, kde = True)
plt.xlabel('Standardized data', fontsize=14)
plt.ylabel('Frequency', fontsize=12)
plt.title("Apartment rents", fontsize=18)
plt.show()
```

### Data Normalization
- Data is converted linearly to values where the smallest is 0 and largest is 1, i.e. MinMax scaling
- This can be done using MinMaxScaler from the Scikit-Learn library
- fit_transform completes the conversion for the MinMaxScaler object

```python
# data normalization with sklearn
from sklearn.preprocessing import MinMaxScaler
minMaxS = MinMaxScaler()
# transform data
scaler = minMaxS.fit_transform(df)
print(scaler.transpose())
```

```python
# data as it is ...
scaler
```

```python
# Create Serie where normalized data is included
norm_rent = scaler[:,0]
norm_rent
```

#### Plot normalized data using Histogram
- Note that the data shape with histogram is same to the original data.
- Bins = 40 is used just for the example

```python
# plot data
sns.histplot(data=norm_rent ,bins=40, kde=True)
plt.xlabel('Normalized data', fontsize=14)
plt.ylabel('Frequency', fontsize=12)
plt.title("Apartment rents", fontsize=18)
plt.show()
```

# My Reflections

1. StandardScaler uses the mean and standard deviation, so its values normally range from -3 to 3. Other values are treated as outliers.
2. MinMaxScaler has a range of values from 0 to 1. If the data has outliers, the whole thing will be stretched by the outliers.
3. RobustScaler uses the median and IQR (Interquartile Range, indicating the range from 25% to 75%). This is the best approach to keep the output data of the best quality.

## Simple Linear Regression
- one independent (x) and one dependent (y) variable
- use LinearRegression from the Scikit-Learn library

```python
# Take Linear regression in use from Scikit-learn
from sklearn.linear_model import LinearRegression
# import scipy.stats as stats # library for statistical measurementssns.get_dataset_names()
```

```python
# Check names of sns datasets
sns.get_dataset_names()
```

```python
# Load data
df_tips = sns.load_dataset('tips')
df_tips.head()
```

```python
df_tips.info()
```

```python
# create vectors containing the data
# x independent variable
x = df_tips['total_bill'].values
x
```

### NOTE: Above x is a normal array, and therefore need to transformed to 1-dimensional ndarray

```python
x.shape
```

```python
# Create vector x containing the data 'total_bill' as 1-dimensional ndarray
x = df_tips['total_bill'].values.reshape(-1,1)
print("shape:", x.shape)
# show 10 first items of the array x
print("first 10 values of x:", x [0:10])
```

```python
# Show the last 10 values of the array x using slicing
print("last 10 values of x:", x [-10:])
```

```python
# create vector y containing the data 'tip'
y = df_tips['tip'].values
# show 20 first items of the array y
y [0:20]
```

```python
# Create linear regression model
model = LinearRegression()
# fit the data
model.fit(x, y)
# calculate the coefficients
slope = model.coef_[0]
intercept = model.intercept_
print(f"Slope: {slope}")
print(f"Intercept: {intercept}")
```

## Visualise the results using a scatter plot

```python
# Visualise the results
# show the scatter diagram for the data and the linear regression line
plt.scatter(x, y, label='Data')
plt.plot(x, model.predict(x), color='red', label='Regression line')
plt.xlabel('x = total bill')
plt.ylabel('y = tip')
plt.title('Relationship between bills and tips')
plt.legend()
plt.show()
```

```python
# Add the equation formula as a text annotation to the plot
plt.scatter(x, y, label='Data')
plt.plot(x, model.predict(x), color='red', label='Regression line')
# Add the equation formula as a text annotation to the plot
equation_text = f'y = {slope:.2f} * x + {intercept:.2f}'
plt.text(2, 7, equation_text, fontsize=12, color='black')
plt.xlabel('x = total bill')
plt.ylabel('y = tip')
plt.title('Relationship between bills and tips')
plt.legend()
plt.show()
```

### Note: Figure could have been created also with Seaborn lmplot
- lmplot uses orginal data from DataFrame and adds the regression line

```python
# Two quantitative variables, one in x and another in y
sns.lmplot(data = df_tips, x= 'total_bill', y= 'tip')
plt.xlabel('x = total bill', fontsize=12)
plt.ylabel('y = tip', fontsize=18)
plt.show()
```

### NICE TO KNOW INFORMATION: Intercept and coefficient can be also found using library statsmodel

```python
# Create figures using statsmodel
fig = plt.figure(figsize=(14, 8))
fig = sm.graphics.plot_regress_exog(model,'total_bill', fig = fig)
plt.show()
```

## Multiple linear regression
- more than one independent variable (X) and one dependent (y) variable
- use LinearRegression from the Scikit-Learn library

```python
# Load data
df_diamonds = sns.load_dataset('diamonds')
df_diamonds.head()
```

```python
df_diamonds.info()
```

```python
df_diamonds.shape
```

```python
# Reduce the size of the data (just for the sake of this example, usually
# reducing the data shall not be done without carefull consideration why and how
# to do it, but let's leave that out of scope of this course now.)
df1 = df_diamonds.loc [0:10000, 'carat':'price']
df1.shape
```

```python
df1.columns
```

```python
# use carat and depth as independent variables store them in matrix X
# use price as dependent variable y
X = df1[['carat', 'depth']]
y = df1['price']
```

```python
model = LinearRegression()
model.fit(X, y)

coefficients = model.coef_
intercept = model.intercept_

print("Coefficients:", coefficients)
print("Intercept:", intercept)
```

### Visualise the results in 3D Scatter Diagram

```python
# Visualise the results in 3D
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x1 = df1['carat']
x2 = df1['depth']
y = df1['price']
ax.scatter(x1, x2, y, c='green', marker='o')
ax.set_xlabel('carat')
ax.set_ylabel('depth')
ax.set_zlabel('price')
plt.title('3D Scatter Plot')
plt.show()
```

```python
# Add the regression plane in the figure
# Extract the coefficients
coef_X1, coef_X2 = model.coef_
intercept = model.intercept_
# create the figure
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x1 = df1['carat']
x2 = df1['depth']
y = df1['price']
# Create a meshgrid for the plane
x1_plane, x2_plane = np.meshgrid(x1, x2)
y_plane = intercept + coef_X1 * x1_plane + coef_X2 * x2_plane
ax.scatter(x1, x2, y, c='blue', marker='o', label='Data Points')
# plot the regression plane
ax.plot_surface(x1_plane, x2_plane, y_plane, alpha=0.5,
                cmap='viridis', label='Regression Plane')
ax.set_xlabel('carat')
ax.set_ylabel('depth')
ax.set_zlabel('price')
plt.title('3D Scatter Plot with Regression Plane')
plt.show()
```

### Tasks, multiple linear regression

Use Seaborn dataset 'planets'
- Print out the basic information.
- Drop rows having missing values, check that it worked.
- Find multiple linear reqression between two independent variables (mass, distance) and a dependent variable (orbital_period).
- Print out intercept and coefficients.
- Create a 3D figure that shows also regression plane.

```python
# Your code here
planets = sns.load_dataset('planets')
planets.info()
```

```python
planets.dropna(inplace=True)
planets.info()
```

```python
model = LinearRegression()
X = planets[['distance', 'mass']]
y = planets['orbital_period']
model.fit(X, y)
```

```python
print("Intercept:", model.intercept_)
print("Coefficients (mass, distance):", model.coef_)
```

```python
# I used AI to support a better drawing performance here

x1 = planets['distance']      # coef_[0]
x2 = planets['mass']          # coef_[1]
y = planets['orbital_period']

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(x1, x2, y, c='blue', marker='o', alpha=0.5, label='Data Points')

x1_range = np.linspace(x1.min(), x1.max(), 30)
x2_range = np.linspace(x2.min(), x2.max(), 30)
X1, X2 = np.meshgrid(x1_range, x2_range)

Y_plane = (model.intercept_
           + model.coef_[0] * X1   # distance
           + model.coef_[1] * X2)  # mass

ax.plot_surface(X1, X2, Y_plane, alpha=0.5, cmap='viridis')

ax.set_xlabel('distance')
ax.set_ylabel('mass')
ax.set_zlabel('orbital_period')
plt.title('3D Scatter Plot with Regression Plane')
plt.show()
```
