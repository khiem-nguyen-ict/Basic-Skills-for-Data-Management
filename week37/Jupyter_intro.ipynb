## Plotting quantitative data
- Histogram is used to show the distribution of quantitative data
- plot.hist using Matplotlib (https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html)
- sns.histplot using Seaborn (https://seaborn.pydata.org/generated/seaborn.histplot.html)

```python
plt.hist(trips.Age, bins=10, color='r',edgecolor='blue')   
plt.xlabel('Age', fontsize=16)
plt.ylabel("Freq")
plt.title("Matplotlib: Histogram of travellers' age", fontsize=18)
plt.show() 
```

### Histogram with Seaborn
- Seaborn is another library for graphics
- Based on Matplotlib
- histplot is a function in Seaborn for Historgrams 
- https://seaborn.pydata.org/generated/seaborn.histplot.html

```python
sns.histplot(data = trips, x= 'Age', color='yellow', 
             edgecolor ='red',bins=10, hue = 'Gender') 
plt.xlabel('Age', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of travellers' age per Gender", fontsize=18)
plt.show()
```

##### Note: Using parameter multiple you get separate bars per variable you have in hue parameter

```python
sns.histplot(data = trips, x= 'Age', color='yellow', 
             edgecolor ='red',bins=10, hue = 'Gender', multiple = 'dodge') 
plt.xlabel('Age', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of travellers' age per Gender", fontsize=18)
plt.show()
```

Modify the histogram's parameters and labels in order to see, what do they do.

```python
# Modify some parameters

sns.histplot(data = trips, x= 'Trips_year', color='blue', 
             edgecolor ='green',bins=3, hue = 'Gender', multiple = 'dodge') 
plt.xlabel('Age', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of travellers' trip years per Gender", fontsize=18)
plt.show()
```

### Task 3
- Create histograms using the 'tips' data!
- At least two different graphs, one with Matplotlib and one with Seaborn

```python
# Task 3
# Matplotlib — histogram of the quantitative variable 'tip'
plt.hist(tips.tip, bins=10, color='r', edgecolor='blue')
plt.xlabel('tip', fontsize=16)
plt.ylabel("Freq")
plt.title("Matplotlib: Histogram of tip amount", fontsize=18)
plt.show()

# Seaborn — histogram of 'tip' grouped by 'sex'
sns.histplot(data=tips, x='tip', color='blue',
             edgecolor='green', bins=10, hue='sex', multiple='dodge')
plt.xlabel('tip', fontsize=14)
plt.ylabel('Freq', fontsize=12)
plt.title("Seaborn: Histogram of tip amount per sex", fontsize=18)
plt.show()
```

## Plotting Qualitative data

### Bar graphs
- Plot qualitative data

#### Seaborn countplot
- https://seaborn.pydata.org/generated/seaborn.countplot.html

```python
# make bar graph for smokers per day
sns.countplot(data = tips, x='day', hue = 'smoker')
plt.title("Seaborn: Bar Graph for smokers per day", fontsize=18)
plt.show()
```

### Task 4
- Create other bar graphs using the 'tips' data!
- At least two different graphs

```python
# Create other bar grahp
# Matplotlib — bar chart
frequencies = tips.day.value_counts().sort_index() # sort by day

plt.bar(frequencies.index,  frequencies.values, color='green', edgecolor='blue')
plt.xlabel('time', fontsize=16)
plt.ylabel("Freq")
plt.title("Matplotlib: Bar graph for smokers per time", fontsize=16)
plt.show()
# Seaborn
sns.countplot(data=tips, x='day', hue='smoker')
plt.title("Seaborn: Bar graph for smokers per time", fontsize=16)
plt.show()
```
